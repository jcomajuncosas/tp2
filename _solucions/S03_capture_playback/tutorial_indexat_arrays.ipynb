# Mini-tutorial — Llegir trossos d'un array (slicing)

Un repàs curt per dominar com s'agafen parts d'un array amb NumPy: des de l'inici, el final, el mig, invertit… És la base de molts efectes (per exemple, l'eco del delay).

**Com funciona:** primer mira la *taula de referència*; després fes els *exercicis* (escriu l'expressió i executa — l'autotest et diu si és correcta). No cal entregar res: és per practicar.

```python
import numpy as np

# El nostre array de proves: el VALOR coincideix amb la POSICIÓ (índex)
a = np.arange(10)
print('a =', a)
print('posicions:', list(range(10)))
```

## 1. Taula de referència

Executa la cel·la i compara cada expressió amb el seu resultat. Fixa't en la idea clau: en `a[i:j]`, s'inclou `i` però **no** `j`.

```python
exemples = [
    ('a[0]',      a[0]),        # primer element
    ('a[-1]',     a[-1]),       # últim element
    ('a[:3]',     a[:3]),       # els 3 PRIMERS
    ('a[-3:]',    a[-3:]),      # els 3 ÚLTIMS
    ('a[2:]',     a[2:]),       # des de la posició 2 fins al final
    ('a[:-2]',    a[:-2]),      # tot MENYS els 2 últims
    ('a[3:6]',    a[3:6]),      # del mig: posicions 3,4,5
    ('a[::-1]',   a[::-1]),     # INVERTIT
    ('a[::2]',    a[::2]),      # un de cada dos (pas 2)
]
for expr, res in exemples:
    print(f'{expr:10} -> {res}')
```

## 3. El cas de l'eco

A l'echo del delay fèiem servir dues llesques alhora. La clau: **tenen la mateixa llargada**, per això es poden sumar element a element. Executa-ho i comprova-ho.

```python
d = 3   # delay en mostres (exemple petit)

print('a         =', a)
print('a[d:]     =', a[d:],  '  (on ATERRA l\'eco, desplaçat d posicions)')
print('a[:-d]    =', a[:-d], '  (la FONT de l\'eco)')
print()
print('mateixa llargada?', len(a[d:]) == len(a[:-d]), '->', len(a[d:]), 'i', len(a[:-d]))
print()
# Alineació: a[0] va amb a[d], a[1] amb a[d+1]...  (cada mostra reapareix d posicions més tard)
for i in range(len(a[:-d])):
    print(f'  font a[{i}]={a[i]}  ->  aterra a la posició {d+i}')
```
